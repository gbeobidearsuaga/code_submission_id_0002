## Compute eady growth rate via hypsometric equation (we do not have geopotential height)

```python
#layer thickness
Rd = 287.0       # J kg^-1 K^-1
g = 9.80665     # m s^-2

Tbar = 0.5 * (ta_850_ML + ta_700_ML)
dz_ML = (Rd / g) * Tbar * np.log(85000 / 70000)

Tbar = 0.5 * (ta_850_MPI + ta_700_MPI)
dz_MPI = (Rd / g) * Tbar * np.log(85000 / 70000)

Tbar = 0.5 * (ta_850_era + ta_700_era)
dz_era = (Rd / g) * Tbar * np.log(85000 / 70000)
```

```python
#vertical wind shear
du_ML = ua_700_ML - ua_850_ML
dv_ML = va_700_ML - va_850_ML
dV_dz_ML = np.sqrt(du_ML**2 + dv_ML**2) / dz_ML

du_MPI = ua_700_MPI - ua_850_MPI
dv_MPI = va_700_MPI - va_850_MPI
dV_dz_MPI = np.sqrt(du_MPI**2 + dv_MPI**2) / dz_MPI

du_era = ua_700_era - ua_850_era
dv_era = va_700_era - va_850_era
dV_dz_era = np.sqrt(du_era**2 + dv_era**2) / dz_era
```

```python
#statis stability

#potential temperature
p0 = 100000.0
kappa = 0.286

theta_850_ML = ta_850_ML * (p0 / 85000)**kappa
theta_700_ML = ta_700_ML * (p0 / 70000)**kappa

theta_850_MPI = ta_850_MPI * (p0 / 85000)**kappa
theta_700_MPI = ta_700_MPI * (p0 / 70000)**kappa

theta_850_era = ta_850_era * (p0 / 85000)**kappa
theta_700_era = ta_700_era * (p0 / 70000)**kappa

#vertical gradient
dtheta_dz_ML = (theta_700_ML - theta_850_ML) / dz_ML
theta_bar_ML = 0.5 * (theta_700_ML + theta_850_ML)
N_ML = np.sqrt((g / theta_bar_ML) * dtheta_dz_ML)

dtheta_dz_MPI = (theta_700_MPI - theta_850_MPI) / dz_MPI
theta_bar_MPI = 0.5 * (theta_700_MPI + theta_850_MPI)
N_MPI = np.sqrt((g / theta_bar_MPI) * dtheta_dz_MPI)

dtheta_dz_era = (theta_700_era - theta_850_era) / dz_era
theta_bar_era = 0.5 * (theta_700_era + theta_850_era)
N_era = np.sqrt((g / theta_bar_era) * dtheta_dz_era)
```

```python
#coriolis
Omega = 7.2921159e-5
f = 2 * Omega * np.sin(np.deg2rad(ua_ML.lat))
```

```python
# Eady growth rate
sigma_ML = 0.31 * np.abs(f) * dV_dz_ML / N_ML
sigma_MPI = 0.31 * np.abs(f) * dV_dz_MPI / N_MPI
sigma_era = 0.31 * np.abs(f) * dV_dz_era / N_era
```

```python
from scipy.ndimage import gaussian_filter

R_earth = 6371000  # meters

def smooth_sst(sst, sigma=2):
    data = sst.values.copy()
    mask = np.isnan(data)         # True where NaN
    valid = ~mask                 # True where data is valid
    filled = np.where(valid, data, 0.0)

    sigma_tuple = (0,) * (data.ndim - 2) + (sigma, sigma)
    smoothed_data = gaussian_filter(filled, sigma=sigma_tuple, mode='constant', cval=0.0)
    smoothed_mask = gaussian_filter(valid.astype(float), sigma=sigma_tuple, mode='constant', cval=0.0)

    smoothed = smoothed_data / np.where(smoothed_mask > 1e-3, smoothed_mask, np.nan)
    smoothed[mask] = np.nan

    return xr.DataArray(smoothed, coords=sst.coords, dims=sst.dims)

def compute_sst_gradients(sst):
    lat = sst.lat.values
    lon = sst.lon.values

    dlat = np.gradient(np.deg2rad(lat)) * R_earth          # (lat,)
    dlon = np.gradient(np.deg2rad(lon)) * R_earth          # (lon,)
    dx = np.outer(np.cos(np.deg2rad(lat)), dlon)           # (lat, lon)
    dy = np.tile(dlat[:, np.newaxis], (1, len(lon)))       # (lat, lon)

    dSST_dlon, dSST_dlat = np.gradient(sst.values, axis=(-2, -1))

    dSST_dx = dSST_dlon / dx
    dSST_dy = dSST_dlat / dy
    gradSST = np.sqrt(dSST_dx**2 + dSST_dy**2)

    dSST_dx = xr.DataArray(dSST_dx, coords=sst.coords, dims=sst.dims)
    dSST_dy = xr.DataArray(dSST_dy, coords=sst.coords, dims=sst.dims)
    gradSST = xr.DataArray(gradSST, coords=sst.coords, dims=sst.dims)

    return dSST_dx, dSST_dy, gradSST

def multi_scale_gradients(sst, sigma_list=[1, 2, 3]):
    grads = {}
    for sigma in sigma_list:
        sst_smooth = smooth_sst(sst, sigma=sigma)
        grads[sigma] = compute_sst_gradients(sst_smooth)
    return grads

sigma_list = [1, 2, 3]

sst_era_gradients = multi_scale_gradients(sst_era, sigma_list)
sst_MPI_gradients = multi_scale_gradients(sst_MPI, sigma_list)
sst_ML_gradients  = multi_scale_gradients(sst_ML, sigma_list)
```

# Seasonal mean

```python
def compute_seasonal_means(da):
    time = da['time'].to_index()

    season_year = time.year + (time.month == 12)
    season = da['time'].dt.season.values

    season_id = [f"{y}-{s}" for y, s in zip(season_year, season)]

    da = da.assign_coords(
        season=('time', season),
        season_year=('time', season_year),
        season_id=('time', season_id)
    )

    grouped = da.groupby('season_id').mean('time')

    season_ts = {}
    for sid in grouped['season_id'].values:
        year, s = sid.split('-')
        year = int(year)
        if s not in season_ts:
            season_ts[s] = []
        da_season = grouped.sel(season_id=sid).expand_dims({'season_year': [year]})
        season_ts[s].append(da_season)

    for season in season_ts:
        season_ts[season] = xr.concat(season_ts[season], dim='season_year')

    return season_ts
```

```python
# Compute seasonal mean of baroclinicity
sigma_MPI_seasonal = compute_seasonal_means(sigma_MPI)
sigma_ML_seasonal = compute_seasonal_means(sigma_ML)
sigma_era_seasonal = compute_seasonal_means(sigma_era)
```

```python
# Compute seasonal mean of SST gradient
sst_era_gradients_sel_seasonal = compute_seasonal_means(sst_era_gradients_sel)
sst_MPI_gradients_sel_seasonal = compute_seasonal_means(sst_MPI_gradients_sel)
sst_ML_gradients_sel_seasonal = compute_seasonal_means(sst_ML_gradients_sel)
```
