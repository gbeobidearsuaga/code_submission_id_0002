```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Hybrid ML

```python
#compute climatology from assimilation
path = '/work/uo1075/u241308/ML_infilling/sst/'
file ='asseikeraf_sst_r1-16i8p4_360x180.nc'

ref_min = 1985
ref_max = 2014

with xr.open_dataset(path+file) as sst_clim:
    sst_clim = sst_clim.tos.mean(dim='depth') #average over ensemble members
    sst_clim = sst_clim[(sst_clim.time.dt.year>=ref_min)&(sst_clim.time.dt.year<=ref_max)]
    sst_clim = sst_clim.groupby('time.month').mean(dim='time')
```

```python
path_in = '/work/uo1075/u241308/mpiesm-1.2.01p7-levante/proccessed_output/sst/'
path_out = '/work/uo1075/u241308/data_python_PostDoc/ML_assimilation/sst_ML_NA/anomaly/'
y_start = 1958
y_end = 2021

for y_id in range(y_start,y_end):
    file_in = 'sst_%i_r1-16i2p3-LR_26_months_360x180.nc' %y_id
    file_out = 'sst_%i_r1-16i2p3-LR_26_months_360x180_anomaly_ref_%i-%i.nc' %(y_id,ref_min,ref_max)
    with xr.open_dataset(path_in+file_in) as sst:
        sst = sst.tos
    sst_anomaly = sst.groupby('time.month') - sst_clim
    sst_anomaly.to_netcdf(path_out+file_out)
    print('Year %i done'%y_id)
```

# Standard

```python
path_in = '/work/uo1075/u241308/data_python_PostDoc/ML_assimilation/sst_benchmark/processed/'
path_out = '/work/uo1075/u241308/data_python_PostDoc/ML_assimilation/sst_benchmark/anomaly/'
y_start = 1960
y_end = 2020

for y_id in range(y_start,y_end):
    file_in = 'sst_%i_r1-16i2p2-LR_3_years_360x180.nc' %y_id
    file_out = 'sst_%i_r1-16i2p2-LR_3_years_360x180_anomaly_ref_%i-%i.nc' %(y_id,ref_min,ref_max)
    with xr.open_dataset(path_in+file_in,decode_times=False) as sst:
        sst = sst.tos
        #time axis is wrong. we set a new one
        units, reference_date = sst.time.attrs['units'].split('since')
        list1 = list(reference_date)
        list1[9:11]='01'
        reference_date = ''.join(list1)
        sst['time'] = pd.date_range(start=reference_date, periods=sst.sizes['time'], freq='MS')
    sst_anomaly = sst.groupby('time.month') - sst_clim
    sst_anomaly.to_netcdf(path_out+file_out)
    print('Year %i done'%y_id)
```
