```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# Hybrid ML

```python
#compute climatology from assimilation
path = '/work/uo1075/u241308/ML_infilling/hfss/'
file ='asseikeraf_hfss_r1-16i8p4_360x180.nc'

ref_min = 1985
ref_max = 2014

with xr.open_dataset(path+file,decode_times=False) as hfss_clim:
    hfss_clim = hfss_clim.ahfs.mean(dim='sfc') #average over ensemble members
    #Time index is not readable for .dt. We create a new index
    reference_date = str('1958-1-01')
    print("Reference starting date: " + reference_date)
    hfss_clim['time'] = pd.date_range(start=reference_date, periods=hfss_clim.sizes['time'], freq='MS')
    
    hfss_clim = hfss_clim[(hfss_clim.time.dt.year>=ref_min)&(hfss_clim.time.dt.year<=ref_max)]
    hfss_clim = hfss_clim.groupby('time.month').mean(dim='time')
```

```python
path_in = '/work/uo1075/u241308/mpiesm-1.2.01p7-levante/proccessed_output/hfss/'
path_out = '/work/uo1075/u241308/data_python_PostDoc/ML_assimilation/hfss_ML_NA/anomaly/'
y_start = 1958
y_end = 2021

for y_id in range(y_start,y_end):
    file_in = 'hfss_%i_r1-16i2p3-LR_26_months_360x180.nc' %y_id
    file_out = 'hfss_%i_r1-16i2p3-LR_26_months_360x180_anomaly_ref_%i-%i.nc' %(y_id,ref_min,ref_max)
    with xr.open_dataset(path_in+file_in) as hfss:
        hfss = hfss.hfss
    hfss_anomaly = hfss.groupby('time.month') - hfss_clim
    hfss_anomaly.to_netcdf(path_out+file_out)
    print('Year %i done'%y_id)
```

# Standard

```python
path_in = '/work/uo1075/u241308/data_python_PostDoc/ML_assimilation/hfss_benchmark/processed/'
path_out = '/work/uo1075/u241308/data_python_PostDoc/ML_assimilation/hfss_benchmark/anomaly/'
y_start = 1960
y_end = 2020

for y_id in range(y_start,y_end):
    file_in = 'hfss_%i_r1-16i2p2-LR_3_years_360x180.nc' %y_id
    file_out = 'hfss_%i_r1-16i2p2-LR_3_years_360x180_anomaly_ref_%i-%i.nc' %(y_id,ref_min,ref_max)
    with xr.open_dataset(path_in+file_in,decode_times=False) as hfss:
        hfss = hfss.hfss
        #time axis is wrong. we set a new one
        units, reference_date = hfss.time.attrs['units'].split('since')
        list1 = list(reference_date)
        list1[9:11]='01'
        reference_date = ''.join(list1)
        hfss['time'] = pd.date_range(start=reference_date, periods=hfss.sizes['time'], freq='MS')
    hfss_anomaly = hfss.groupby('time.month') - hfss_clim
    hfss_anomaly.to_netcdf(path_out+file_out)
    print('Year %i done'%y_id)
```
